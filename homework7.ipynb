```python
def W7A16():
    arr = list(map(int, input().split()))
    count = Counter(arr)
    to_remove = 0
    result = []
    for x, cnt in count.items():
        if x == 0:
            to_remove += cnt
        elif cnt >= x:
            to_remove += cnt - x
            result.extend([x] * x)
        else:
            to_remove += cnt
    if to_remove == 0:
        print("Mang dep!")
    else:
        print(f"Khong phai mang dep, xoa toi thieu {to_remove} phan tu")
        print("Mang sau khi xoa:", sorted(result))
W7A16()
 
```

```python
def W7A17():
    arr = list(map(int, input().split()))
    candidate, vote = None, 0
    for x in arr:
        if vote == 0:
            candidate, vote = x, 1
        elif x == candidate:
            vote += 1
        else:
            vote -= 1
    if arr.count(candidate) > len(arr) // 2:
        print(candidate)
W7A17()
```

```python
def W7A19():
    def is_asc(a):
        return len(a) < 2 or all(a[i] <= a[i+1] for i in range(len(a)-1))
    arr1 = list(map(int, input().split()))
    arr2 = list(map(int, input().split()))
    a = arr1 if is_asc(arr1) else arr1[::-1]
    b = arr2 if is_asc(arr2) else arr2[::-1]
    result, i, j = [], 0, 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i]); i += 1
        else:
            result.append(b[j]); j += 1
    result.extend(a[i:])
    result.extend(b[j:])
    print(result)
W7A19()
```

```python
def W7A25():
    s = input().strip()
    k = int(input())
    n = len(s)
    result, start = [], 0
    for i in range(k):
        end = n - k + i
        max_digit = max(s[start:end+1])
        idx = s.index(max_digit, start, end+1)
        result.append(max_digit)
        start = idx + 1
    print("".join(result))
W7A25()
 
```

```python
def W7A27():
    s = input().strip()
    stack = []
    pairs = {')': '(', '}': '{', ']': '['}
    for c in s:
        if c in '({[':
            stack.append(c)
        elif c in ')}]':
            if not stack or stack[-1] != pairs[c]:
                print("Khong hop le")
                return
            stack.pop()
    print("Hop le" if not stack else "Khong hop le")
W7A27()
 
```

```python

def W7A28():
    N = int(input())
    def is_prime(n):
        if n < 2: return False
        if n == 2: return True
        if n % 2 == 0: return False
        for i in range(3, int(n**0.5)+1, 2):
            if n % i == 0: return False
        return True
    upper = max(N + 10000, 1000)
    best_start = best_len = curr_len = 0
    curr_start = -1
    for num in range(N+1, upper+1):
        if not is_prime(num):
            if curr_len == 0: curr_start = num
            curr_len += 1
        else:
            if curr_len > best_len:
                best_len, best_start = curr_len, curr_start
            curr_len = 0
    if curr_len > best_len:
        best_len, best_start = curr_len, curr_start
    if best_len > 0:
        print(f"Day hop so dai nhat: {list(range(best_start, best_start + best_len))} (do dai {best_len})")
    else:
        print("Khong tim thay day hop so nao")
W7A28()
```

```python
def W7A29():
    lst = list(map(int, input().split()))
    best_start = curr_start = 0
    best_len = curr_len = 1
    for i in range(1, len(lst)):
        if lst[i] > lst[i-1]:
            curr_len += 1
        else:
            if curr_len > best_len:
                best_len, best_start = curr_len, curr_start
            curr_start = i
            curr_len = 1
    if curr_len > best_len:
        best_len, best_start = curr_len, curr_start
    print(lst[best_start:best_start + best_len])
W7A29()
```

```python
def W7A30():
    lst = list(map(int, input().split()))
    n = len(lst)
    dp = [1] * n
    pre = [-1] * n
    for i in range(1, n):
        for j in range(i):
            if lst[j] < lst[i] and dp[j] + 1 > dp[i]:
                dp[i] = dp[j] + 1
                pre[i] = j
    max_len = max(dp)
    idx = dp.index(max_len)
    result = []
    while idx != -1:
        result.append(lst[idx])
        idx = pre[idx]
    print(result[::-1])
W7A30()
```

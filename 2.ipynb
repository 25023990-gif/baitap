```python
a = float(input())
b = float(input())

r = a / 2
dien_tich_con_lai = (a * b) - (3.14 * (r ** 2))

print(f"{dien_tich_con_lai:.2f}")
```

```python
import math
a = float(input())
b = float(input())
c = float(input())
if a + b >c and a + c > b and b + c > a :
    p = (a + b + c) / 2
    dien_tich = math.sqrt(p * (p - a) * (p - b) * (p - c))
    print(f"{dien_tich:.1f}")
else :
    print(f"Khong phai ba canh cua tam giac")
```

```python
def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
import math

a = float(input("Nhập cạnh a: "))
b = float(input("Nhập cạnh b: "))
c = float(input("Nhập cạnh c: "))

if a + b > c and a + c > b and b + c > a:
    chu_vi = a + b + c
    p = chu_vi / 2
    dien_tich = math.sqrt(p * (p - a) * (p - b) * (p - c))
    
    canh = sorted([a, b, c])
    
    if a == b == c:
        loai = "Tam giác đều"
    elif a == b or b == c or a == c:
        if math.isclose(canh[0]**2 + canh[1]**2, canh[2]**2):
            loai = "Tam giác vuông cân"
        else:
            loai = "Tam giác cân"
    elif math.isclose(canh[0]**2 + canh[1]**2, canh[2]**2):
        loai = "Tam giác vuông"
    else:
        loai = "Tam giác thường"
        
    print(f"Đây là {loai}. Chu vi: {chu_vi}, Diện tích: {dien_tich}")
else:
    print("Khong la ba canh cua mot tam giac")
```
